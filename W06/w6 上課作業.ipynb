```python
import numpy as np
import pandas as pd

# ========== Mission 4：單筆範例（人为公式） ==========
study_hours, sleep_hours, screen_time_hours, attendance_rate = 2.5, 7.0, 4.0, 90
base_score = 50

study_effect = 3 * study_hours              # 3 × 讀書時數
sleep_effect = 2 * sleep_hours              # 2 × 睡眠時數
screen_effect = -1.2 * screen_time_hours    # -1.2 × 螢幕時數
attendance_effect = 0.15 * attendance_rate  # 0.15 × 出席率

assert None not in (study_effect, sleep_effect, screen_effect, attendance_effect), "請先補完四項影響。"

sample_score = base_score
sample_score = sample_score + study_effect
sample_score = sample_score + sleep_effect
sample_score = sample_score + screen_effect
sample_score = sample_score + attendance_effect
print("模擬成績：", round(sample_score, 1))

# ========== 建立 60 筆學生資料 df ==========
n = 60
df = pd.DataFrame({
    "student_id": [f"S{i:03d}" for i in range(1, n + 1)],
    "study_hours": np.round(np.random.default_rng(1).uniform(0.5, 6.0, n), 1),
    "sleep_hours": np.round(np.random.default_rng(2).uniform(4.0, 9.0, n), 1),
    "screen_time_hours": np.round(np.random.default_rng(3).uniform(0.5, 8.0, n), 1),
    "attendance_rate": np.round(np.random.default_rng(4).uniform(60, 100, n), 1),
})

# ========== Mission 8：建立 quiz_score ==========
rng = np.random.default_rng(42)

study_part = 3 * df["study_hours"]
sleep_part = 2 * df["sleep_hours"]              # 2 × 睡眠時數欄位
screen_part = -1.2 * df["screen_time_hours"]    # -1.2 × 螢幕時間欄位
attendance_part = 0.15 * df["attendance_rate"]
random_error = rng.normal(0, 5, size=n)

assert sleep_part is not None and screen_part is not None, "請先補完欄位運算。"

raw_scores = 50 + study_part
raw_scores = raw_scores + sleep_part
raw_scores = raw_scores + screen_part
raw_scores = raw_scores + attendance_part
raw_scores = raw_scores + random_error

# 限制在 0~100、四捨五入、轉整數
df["quiz_score"] = raw_scores.clip(0, 100).round().astype(int)

assert df["quiz_score"].notna().all(), "請先建立 quiz_score。"
assert df["quiz_score"].between(0, 100).all(), "分數應在 0 到 100。"

# ========== 顯示全部 60 筆 ==========
cols = ["student_id", "study_hours", "sleep_hours",
        "screen_time_hours", "attendance_rate", "quiz_score"]
print(df[cols].shape)
with pd.option_context("display.max_rows", 60):
    try:
        display(df[cols].reset_index(drop=True))
    except NameError:                 # 非 Notebook 環境時的備援
        print(df[cols].reset_index(drop=True).to_string())
```
